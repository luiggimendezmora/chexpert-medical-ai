```python
import pandas as pd
import matplotlib.pyplot as plt

train_df = pd.read_csv('../data/archive/train.csv')

# Check label distribution
pathologies = ['Atelectasis', 'Cardiomegaly', 'Consolidation', 'Edema', 'Pleural Effusion']
print(train_df[pathologies].value_counts())

# Check demographics
print(train_df['Sex'].value_counts())
print(train_df['Age'].describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

pathologies = ['Atelectasis', 'Cardiomegaly', 'Consolidation', 'Edema', 'Pleural Effusion']

# Distribución de etiquetas por patología
label_counts = train_df[pathologies].apply(lambda x: x.value_counts()).T
label_counts.columns = ['Negative (0)', 'Uncertain (-1)', 'Positive (1)']
label_counts = label_counts.fillna(0)

label_counts.plot(kind='bar', figsize=(12, 5), colormap='coolwarm')
plt.title('Label Distribution per Pathology')
plt.ylabel('Count')
plt.xlabel('Pathology')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Distribución por sexo
plt.figure(figsize=(5, 4))
train_df['Sex'].value_counts().plot(kind='bar', color=['steelblue', 'salmon', 'gray'])
plt.title('Patient Distribution by Sex')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
